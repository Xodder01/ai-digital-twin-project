```python
# =========================
# 1. Import Libraries
# =========================
import pandas as pd
import joblib

# =========================
# 2. Load Saved Files
# =========================
model = joblib.load("productivity_model.pkl")
scaler = joblib.load("scaler.pkl")
columns = joblib.load("columns.pkl")  # IMPORTANT

# model = joblib.load("burnout_model.pkl")
# scaler= joblib.load("scaler.pkl")
# columns= joblib.load("columns.pkl")


# =========================
# 3. New User Input
# =========================
# (This will come from frontend later)

user_input = {
    "age": 21,
    "study_hours": 5,
    "self_study_hours": 3,
    "online_classes_hours": 2,
    "social_media_hours": 2,
    "gaming_hours": 1,
    "screen_time_hours": 4,
    "sleep_hours": 6,
    "exercise_minutes": 30,
    "caffeine_intake_mg": 0,
    "mental_health_score": 0,
    "gender": "Male",
    "academic_level": "UG",
    "part_time_job": "No",
    "internet_quality": "Good",
    "upcoming_deadline": "Yes"
}

# =========================
# 4. Convert to DataFrame
# =========================
input_df = pd.DataFrame([user_input])

# =========================
# 5. Apply One-Hot Encoding
# =========================
input_df = pd.get_dummies(input_df)

# =========================
# 6. Match Training Columns
# =========================
input_df = input_df.reindex(columns=columns, fill_value=0)

# =========================
# 7. Scale Input
# =========================
input_scaled = scaler.transform(input_df)
print(input_df)
# =========================
# 8. Predict
# =========================
prediction = model.predict(input_scaled)
burn_out = model.predict(input_scaled)
print("📈 Predicted Productivity Score:", round(prediction[0], 2))
# print("📈 Predicted Burnout Score:", round(burn_out[0], 2))
```
